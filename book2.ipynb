```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 100)
```

```python
def decay_function(x, a, b):
    return a * np.exp(-b * x)
```

```python
a = 5
b = 0.5

y = decay_function(x, a, b)
```

```python
sigma = np.linspace(2, 0.2, len(x))
noise = np.random.normal(0, sigma)

y_scatter = y + noise
```

```python
plt.errorbar(x, y_scatter, yerr=sigma, fmt='o', color='red', ecolor ='black', capsize=3)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Exponential Decay with Scatter Plot")

plt.savefig("question2.pdf")

plt.show()
```
