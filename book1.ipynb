```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 100)
```

```python
def decay_function(x, a, b):
    return a * np.exp(-b * x)
```

```python
a = 5
b = 0.5

y = decay_function(x, a, b)
```

```python
plt.plot(x, y)

plt.xlabel("x")
plt.ylabel("y = a*exp(-b*x)")
plt.title("Exponential Decay Function")

plt.savefig("exp_decay.pdf")

plt.show()
```
