# Sine and Cosine Functions

This notebook defines functions for calculating sin(x) and cos(x)
using NumPy. The values are tabulated between 0 and 2 with
1000 entries.

```python
import numpy as np

def my_sin(x):
    """
    Returns the sine of x.
    """
    return np.sin(x)
```

```python
def my_cos(x):
    """
    Returns the cosine of x.
    """
    return np.cos(x)
```

## Tabulating Values

This cell creates 1000 x-values between 0 and 2,
then computes sin(x) and cos(x).

```python
# Create x values
x = np.linspace(0, 2, 1000)

# Compute sine and cosine values
sin_values = my_sin(x)
cos_values = my_cos(x)
```

## Printing the First 10 Values

This cell prints the first 10 values of x, sin(x),
and cos(x) in column format.

```python
# Print table header
print(f"{'x':>12} {'sin(x)':>12} {'cos(x)':>12}")

# Print first 10 rows
for i in range(10):
    print(f"{x[i]:12.6f} {sin_values[i]:12.6f} {cos_values[i]:12.6f}")
```
